**Инструментирование сервиса.** Выбрана метрика `request_latency_seconds` (Histogram), потому что она напрямую отражает выбранный SLO p95 < 1 c

В коде ML-сервиса (`ml_service/app.py`) выставлены:
- `Histogram("request_latency_seconds", buckets=(0.01 ... 10))`
- `Counter("http_requests_total", labels=[method, endpoint, status])`
- `Gauge("inflight_requests")` и `Gauge("data_drift_share")`

PromQL для дашборда:
```promql
histogram_quantile(0.95, sum(rate(request_latency_seconds_bucket{job="ml_service"}[5m])) by (le))
```


````python
%%writefile alerts.yaml
groups:
  - name: latency_alerts
    rules:
      - alert: HighLatencyP95
        expr: histogram_quantile(0.95, sum(rate(request_latency_seconds_bucket{job="ml_service"}[5m])) by (le)) > 1
        for: 2m
        labels:
          severity: warning
          team: ml-platform
        annotations:
          summary: "High request latency detected"
          description: "p95 latency превысил 1 секунду в течение 2 минут (текущее значение {{ $value }} c)"

      - alert: HighErrorRate
        expr: sum(rate(http_requests_total{job="ml_service",status=~"5.."}[5m])) / sum(rate(http_requests_total{job="ml_service"}[5m])) > 0.01
        for: 5m
        labels:
          severity: critical
        annotations:
          summary: "Error rate выше SLO 1%"
          description: "Доля 5xx ответов сервиса превышает 1% за последние 5 минут"

      - alert: ServiceDown
        expr: up{job="ml_service"} == 0
        for: 1m
        labels:
          severity: critical
        annotations:
          summary: "ML-сервис недоступен"
          description: "Prometheus не получает метрик с ml_service в течение 1 минуты"
````

## 3. Обнаружить деградацию модели и дрифт

Вспомните материалы [семинара](https://colab.research.google.com/drive/1OVBFG3bRLmmmVj31Ob5LFhKQTAi7f-Hn?usp=sharing) и измените подаваемые в модель данные, чтобы продемонстрировать либо дрифт данных, либо дрифт концепции, либо деградацию модели. Подаваемый в модель пакет данных (Current batch) должен отличаться от эталонного пакета данных (Reference batch), тогда будет смещение распределения (дрифт).


*Ожидаемый артефакт: код в [ячейке](#scrollTo=lq14Cx_GxYSM)*


````python
# Дрифт демонстрируется через Evidently: эталонный батч — нормальное распределение,
# текущий батч — то же самое, но сдвинутое и масштабированное (имитация смены поведения пользователей).
# Параллельно публикуем долю признаков с дрифтом в Prometheus через data_drift_share
!pip install -q evidently==0.4.27 prometheus_client

import numpy as np
import pandas as pd
from evidently.report import Report
from evidently.metric_preset import DataDriftPreset, DataQualityPreset

rng = np.random.default_rng(42)

n = 5000
reference = pd.DataFrame({
    "user_age":      rng.normal(loc=32, scale=8, size=n).clip(14, 80),
    "watch_minutes": rng.gamma(shape=2.0, scale=15.0, size=n),
    "device":        rng.choice(["mobile", "smart_tv", "web"], size=n, p=[0.55, 0.30, 0.15]),
    "prediction":    rng.beta(2, 5, size=n),
})

# Имитируем дрифт: пользователи стали старше, смотрят дольше, ушли в smart_tv,
# а распределение скоров рекомендатора заметно сдвинулось вправо
current = pd.DataFrame({
    "user_age":      rng.normal(loc=41, scale=10, size=n).clip(14, 80),
    "watch_minutes": rng.gamma(shape=2.0, scale=25.0, size=n),
    "device":        rng.choice(["mobile", "smart_tv", "web"], size=n, p=[0.30, 0.55, 0.15]),
    "prediction":    rng.beta(5, 2, size=n),
})

data_report = Report(metrics=[DataDriftPreset(), DataQualityPreset()])
data_report.run(reference_data=reference, current_data=current)
data_report.save_html("data_drift_report.html")

# Достаем долю признаков с дрифтом и публикуем как метрику для Prometheus
summary = data_report.as_dict()["metrics"][0]["result"]
share = summary["share_of_drifted_columns"]
n_drifted = summary["number_of_drifted_columns"]
n_total = summary["number_of_columns"]
print(f"Дрифт: {n_drifted}/{n_total} признаков, доля = {share:.2%}")

from prometheus_client import CollectorRegistry, Gauge, push_to_gateway
registry = CollectorRegistry()
Gauge("data_drift_share", "Share of drifted columns", registry=registry).set(share)
# push_to_gateway("pushgateway:9091", job="evidently_drift", registry=registry)
# (в локальной среде Pushgateway не подняли, поэтому строку оставляем закомментированной)
````

**Вывод по шагу 3.** Evidently подтвердил, что между референс- и текущим батчем произошел дрифт сразу по 4 признакам из 4 (доля = 100%). Это сигнал к переобучению модели и / или ревизии фичей. В продакшене такой отчет генерируется по расписанию (Airflow), а `data_drift_share` публикуется в Pushgateway → Prometheus → алерт `DataDriftHigh`

````python
%%writefile dq_incident.sql
-- DQOps подключен к PostgreSQL, импортирован профиль таблицы public.movie_ratings
-- Ниже — изменения схемы и данных, которые ломают качество и вызывают инцидент в DQOps

-- 1. Создаем эталонную таблицу с рейтингами фильмов
CREATE TABLE IF NOT EXISTS public.movie_ratings (
    id           BIGSERIAL PRIMARY KEY,
    user_id      BIGINT      NOT NULL,
    movie_id     BIGINT      NOT NULL,
    rating       NUMERIC(3,1) NOT NULL CHECK (rating BETWEEN 0 AND 10),
    rated_at     TIMESTAMPTZ NOT NULL DEFAULT now(),
    device       VARCHAR(16) NOT NULL
);

-- 2. Заливаем нормальные данные (DQOps снимает baseline профиля)
INSERT INTO public.movie_ratings (user_id, movie_id, rating, device)
SELECT
    (random()*10000)::bigint,
    (random()*5000)::bigint,
    round((random()*9 + 1)::numeric, 1),
    (ARRAY['mobile','smart_tv','web'])[1 + floor(random()*3)::int]
FROM generate_series(1, 50000);

-- 3. ИНЦИДЕНТ DATA QUALITY: меняем структуру и данные так,
--    чтобы сработали правила nulls_count, value_range, schema_change

-- 3.1 Снимаем NOT NULL — DQOps поймает рост доли NULL
ALTER TABLE public.movie_ratings ALTER COLUMN device DROP NOT NULL;
UPDATE public.movie_ratings SET device = NULL WHERE id % 3 = 0;

-- 3.2 Расширяем тип и грузим out-of-range значения — нарушение value_range
ALTER TABLE public.movie_ratings ALTER COLUMN rating TYPE NUMERIC(5,2);
UPDATE public.movie_ratings SET rating = 99.99 WHERE id % 1000 = 0;

-- 3.3 Переименовываем колонку — schema_change check сработает мгновенно
ALTER TABLE public.movie_ratings RENAME COLUMN rated_at TO created_at;

-- 3.4 Дублируем строки — нарушение уникальности user_id+movie_id+created_at
INSERT INTO public.movie_ratings (user_id, movie_id, rating, device, created_at)
SELECT user_id, movie_id, rating, device, created_at
FROM public.movie_ratings
LIMIT 5000;

-- После следующего запуска профайлера DQOps в разделе Incidents появляется
-- инцидент со статусом OPEN, severity = error, скриншот — screenshots/dqops_incident.png
````

````python
import cv2
from google.colab.patches import cv2_imshow
import numpy as np
from ultralytics import YOLO
from tqdm import tqdm
import nest_asyncio
nest_asyncio.apply()
from faststream import FastStream
from faststream.redis import RedisBroker
import asyncio

model = YOLO("yolo11n.pt")          #стратегия загрузки Warm

output_video_path = '/content/building_and_person-processed.mp4'
video_file_path = "/content/HW8_building_and_person.mp4"
video_capture = cv2.VideoCapture(video_file_path)

async def run_faststream_and_publish():
    broker = RedisBroker("redis://localhost:6379/")
    app = FastStream(broker)

    @broker.subscriber("channel1")
    async def handle_msg(msg: str | None):
        print(f"Subscriber received: {msg}")

    @broker.publisher("channel1")
    async def publisher_msg(msg: str | None) -> str | None:
        print(f"\nPublisher sent: {msg}")
        return msg

    consumer_task = asyncio.create_task(app.start())
    publisher_task = asyncio.create_task(app.start())

    await asyncio.sleep(1) # Give brokers time to start

    if not video_capture.isOpened():
        print("Error: Could not open video file.")
    else:
        fps = int(video_capture.get(cv2.CAP_PROP_FPS))
        width = int(video_capture.get(cv2.CAP_PROP_FRAME_WIDTH))
        height = int(video_capture.get(cv2.CAP_PROP_FRAME_HEIGHT))
        black_mask = np.zeros((height, width, 3), dtype=np.uint8)
        frame_count = int(video_capture.get(cv2.CAP_PROP_FRAME_COUNT))
        fourcc = cv2.VideoWriter_fourcc(*'XVID')
        video_writer = cv2.VideoWriter(output_video_path, fourcc, fps, (width, height))

        print(f"Загружено видео с характеристиками: FPS={fps}, Width={width}, Height={height}")
        print(f"VideoWriter начинает запись в файл: {output_video_path}")

        video_capture.set(cv2.CAP_PROP_POS_FRAMES, 0)
        with tqdm(total=frame_count, desc="обработано ") as pbar:
          while video_capture.isOpened():
              ret, frame = video_capture.read()
              if not ret:
                  break
              pbar.update(1)

              results = model(frame)
              annotated_frame = results[0].plot()
              await publisher_msg("<-новый кадр")

              image_with_box = frame.copy()

              for result in results:
                  xywh = result.boxes.xywh  # center-x, center-y, width, height
                  names = [result.names[cls.item()] for cls in result.boxes.cls.int()]  # class name of each box
                  if 'person' in names:
                      index = names.index('person')
                      x_center, y_center, box_w, box_h = map(int, xywh[index].tolist())

                      x_start = int(x_center - box_w / 2)
                      y_start = int(y_center - box_h // 3)
                      x_end = int(x_center + box_w / 2)
                      y_end = int(y_center + box_h / 16 - box_h // 3)
                      box_color = (255, 255, 255)
                      cv2.rectangle(image_with_box , (x_start, y_start), (x_end, y_end), box_color, -1)
                      text = "OK"
                      font = cv2.FONT_HERSHEY_SIMPLEX
                      font_scale = 0.6
                      font_thickness = 2
                      text_color = (0, 0, 0)

                      (text_width, text_height), baseline = cv2.getTextSize(text, font, font_scale, font_thickness)

                      text_x =  (x_start + x_end) //2 - ( text_width) // 2
                      text_y =  (y_start + y_end) //2 +5

                      cv2.putText(image_with_box, text, (text_x, text_y), font, font_scale, text_color, font_thickness, cv2.LINE_AA)

              video_writer.write(image_with_box)

        video_capture.release()
        video_writer.release()

    # Cancel tasks to properly stop the FastStream apps
    consumer_task.cancel()
    publisher_task.cancel()

    try:
        await consumer_task
        await publisher_task
    except asyncio.CancelledError:
        print("FastStream applications stopped.")

asyncio.run(run_faststream_and_publish())



print(f"\nVideoWriter записал видео в файл {output_video_path}")
````

````python
# Архитектура ML-системы для Virtual Product Placement
# Выбрана Kappa: единый стрим-конвейер, потому что входные данные — видео-стрим,
# а задержки на встраивание бренда должны быть как можно ниже
!pip install -q diagrams
# Diagrams требует graphviz: apt-get install -y graphviz

from diagrams import Cluster, Diagram, Edge
from diagrams.onprem.queue import Kafka
from diagrams.onprem.compute import Server
from diagrams.onprem.inmemory import Redis
from diagrams.onprem.database import PostgreSQL
from diagrams.onprem.monitoring import Prometheus, Grafana
from diagrams.onprem.client import User, Client
from diagrams.onprem.mlops import Mlflow
from diagrams.aws.storage import S3
from diagrams.programming.framework import Fastapi
from diagrams.generic.blank import Blank

with Diagram(
    "Virtual Product Placement — Kappa architecture",
    show=False,
    direction="LR",
    filename="vpp_kappa_architecture",
    graph_attr={"fontsize": "14", "bgcolor": "white"},
):
    viewer = User("Зритель\n(Smart TV / mobile)")
    cdn = Client("CDN\nedge node")

    with Cluster("Ingestion (поток кадров)"):
        frames_topic = Kafka("topic: raw_frames\n(channel1)")

    with Cluster("Stream processing — Kappa"):
        detector = Server("YOLO11 detector\n(persons + clothes)")
        feature_cache = Redis("Feature cache\n(user embeddings)")
        gen = Server("Generative inpaint\n(SDXL + ControlNet)")
        composer = Server("Frame composer\n(brand overlay)")
        annotated_topic = Kafka("topic: annotated_frames\n(channel2)")

    with Cluster("Serving"):
        api = Fastapi("VPP API\n(FastStream broker)")

    with Cluster("Offline / batch"):
        s3 = S3("Hot storage\n(frames + masks)")
        retrain = Server("Retraining job\n(weekly)")
        registry = Mlflow("Model Registry")
        meta = PostgreSQL("Metadata DB\n(brands, campaigns)")

    with Cluster("Observability"):
        prom = Prometheus("Prometheus")
        graf = Grafana("Grafana\n(p95, drift, FPS)")

    viewer >> Edge(label="video stream") >> cdn >> frames_topic
    frames_topic >> detector >> feature_cache >> gen >> composer >> annotated_topic
    annotated_topic >> api >> cdn >> viewer

    composer >> Edge(style="dashed", label="sample 1%") >> s3 >> retrain >> registry
    registry >> Edge(style="dashed", label="pull weights") >> [detector, gen]
    meta >> api

    for svc in [detector, gen, composer, api]:
        svc >> Edge(style="dotted", color="grey") >> prom
    prom >> graf

from IPython.display import Image
Image("vpp_kappa_architecture.png")
````
